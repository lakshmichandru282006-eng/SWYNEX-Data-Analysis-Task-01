```python

import pandas as pd
import numpy as np

df = pd.read_csv(
    '/content/Raw_Dataset.csv',
    dtype={'PatientId': 'string', 'AppointmentID': 'string'}
)

print("Original Dataset Shape:", df.shape)
display(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

for col in df.select_dtypes(include=['object', 'string']).columns:
    df[col] = df[col].str.strip()

print("Column names and text values standardized.")
print(df.columns.tolist())
```

```python
df['scheduledday'] = pd.to_datetime(
    df['scheduledday'], errors='coerce', utc=True
)

df['appointmentday'] = pd.to_datetime(
    df['appointmentday'], errors='coerce', utc=True
)

df['age'] = pd.to_numeric(df['age'], errors='coerce')

invalid_ages = (df['age'] < 0).sum()
df.loc[df['age'] < 0, 'age'] = np.nan

print("Invalid negative ages corrected:", invalid_ages)
print("Invalid ScheduledDay dates:", df['scheduledday'].isna().sum())
print("Invalid AppointmentDay dates:", df['appointmentday'].isna().sum())
```

```python
duplicates = df.duplicated().sum()

print("Duplicate rows found:", duplicates)

df = df.drop_duplicates().copy()

print("Dataset shape after duplicate check:", df.shape)
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nAge Summary:")
print(df['age'].describe())

print("\nAttendance Categories:")
print(df['no-show'].value_counts(dropna=False))
```

```python
df.to_csv('/content/Cleaned_Dataset.csv', index=False)

print("Cleaned dataset saved successfully.")

from google.colab import files
files.download('/content/Cleaned_Dataset.csv')
```
